```python
from mutagen.mp3 import MP3
from mutagen.id3 import ID3, TIT2, TPE1, TALB, APIC

# 1. Ajusta aquí el nombre de tu archivo subido
file_path = "Love.mp3"  # Cambia esto por el nombre real de tu archivo

try:
    # Cargar el archivo
    audio = MP3(file_path, ID3=ID3)

    # 2. Modificar Metadatos
    audio.tags.add(TIT2(encoding=3, text="Love"))
    audio.tags.add(TPE1(encoding=3, text="Zoé"))
    audio.tags.add(TALB(encoding=3, text="Rocanlover"))

    # Guardar cambios
    audio.save()

    # 3. Mostrar resultados corregidos
    print("--- Información Actualizada ---")
    print(f"Título: {audio.tags.get('TIT2')}")
    print(f"Artista: {audio.tags.get('TPE1')}")
    print(f"Duración: {audio.info.length:.2f} segundos")

except FileNotFoundError:
    print("❌ Error: No encontré el archivo. Asegúrate de haberlo subido a Colab.")
except Exception as e:
    print(f"❌ Ocurrió un error: {e}")
```

```python
from mutagen.mp3 import MP3
from mutagen.id3 import ID3, APIC

# Path to your MP3 file
file_path = "Love.mp3"

# Load the MP3 file
audio = MP3(file_path, ID3=ID3)

# Find and extract the album art (APIC frame)
for tag in audio.tags.values():
    if isinstance(tag, APIC):
        # Write the image data to a file
        with open("album_art.jpg", "wb") as img_file:
            img_file.write(tag.data)
        print("Album art extracted successfully!")
        break
```

```python
from mutagen.mp3 import MP3
from mutagen.id3 import ID3, TIT2, TPE1, TALB

# Path to your MP3 file
file_path = "Love.mp3"

# Load the MP3 file
audio = MP3(file_path, ID3=ID3)

# Extract duration (in seconds)
duration = audio.info.length

# Extract ID3 tags (metadata)
tags = audio.tags

# Extract metadata if available
title = tags.get('TIT2', 'Love').text[0]  # Song title
artist = tags.get('TPE1', 'Zoé').text[0]  # Artist name
album = tags.get('TALB', 'Rocanlover').text[0]  # Album name

# Print metadata
print(f"Title: {title}")
print(f"Artist: {artist}")
print(f"Album: {album}")
print(f"Duration: {duration:.2f} seconds")
```
